## Run All Tests

Execute every test in the repository.

```python
all_results = runner.run_all(verbose=False)

print("\n" + "="*50)
print("  SUMMARY")
print("="*50)
for key, result in all_results.items():
    if isinstance(result, dict):
        passed = result.get('passed', '?')
        status = '✓' if passed else '✗'
        # Handle nested results (lemmas, obstructions)
        if 'lemma5' in result:
            l5 = '✓' if result['lemma5'].get('passed') else '✗'
            l6 = '✓' if result['lemma6'].get('passed') else '✗'
            print(f"  {key:<20s} Lemma5={l5} Lemma6={l6}")
        elif 'ceiling' in result:
            c = '✓' if result['ceiling'].get('passed') else '✗'
            g = '✓' if result['growth'].get('passed') else '✗'
            print(f"  {key:<20s} Ceiling={c} Growth={g}")
        else:
            print(f"  {key:<20s} {status}")

print("\nAll tests complete.")
```
